VERIFICAR SI SE PUEDE USAR LA GPU DE LA MAQUINA DE MI ENTORNO ANACONDA

```python
import torch

# Comprueba si CUDA (la tecnología de NVIDIA para la GPU) está disponible
if torch.cuda.is_available():
    print("¡La GPU está disponible!")
    # Imprime el nombre de la tarjeta gráfica
    print(f"Nombre de la GPU: {torch.cuda.get_device_name(0)}")
    # Mueve un tensor a la GPU para confirmar
    x = torch.randn(10).to("cuda")
    print("Tensor movido a la GPU:", x.is_cuda)
else:
    print("La GPU no está disponible. Se usará la CPU.")
```

VERIFICAR GRAFICAS

```python
import matplotlib.pyplot as plt
import numpy as np

# Genera datos de ejemplo
x = np.linspace(0, 10, 100)
y = np.sin(x)

# Crea la figura y los ejes para la gráfica
plt.figure(figsize=(8, 6))

# Dibuja la función seno
plt.plot(x, y, label='Función Seno')

# Añade título y etiquetas a los ejes
plt.title('Gráfica de Verificación')
plt.xlabel('Eje X')
plt.ylabel('Eje Y')

# Muestra la leyenda y la cuadrícula
plt.legend()
plt.grid(True)

# Muestra la gráfica
plt.show()
```
